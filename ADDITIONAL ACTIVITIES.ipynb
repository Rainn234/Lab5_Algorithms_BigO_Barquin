```python
# --- Student Configuration Setup ---
LAST_NAME = "Barquin"
STUDENT_ID = "TUPM-26-1221"
FAVORITE_ARTIST = "Bruno Mars"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST.replace(" ", ""))
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha())

# --- Functions ---
def rotated_binary_search(arr, target):
    low, high = 0, len(arr) - 1
    comparisons = 0
    
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if arr[mid] == target:
            return mid, comparisons
        
        if arr[low] <= arr[mid]:
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
                
    return -1, comparisons

def linear_search(arr, target):
    comparisons = 0
    for i, val in enumerate(arr):
        comparisons += 1
        if val == target:
            return i, comparisons
    return -1, comparisons

# 1. Dataset Generation
original_dataset = sorted([(SEED_DIGIT + i * 3) for i in range(15)])
rotation_point = (SEED_DIGIT % len(original_dataset)) + 1
rotated_dataset = original_dataset[rotation_point:] + original_dataset[:rotation_point]
target_incident = original_dataset[(ID_SUM + ARTIST_SUM) % len(original_dataset)]

# 2. Search Execution
bin_idx, bin_comp = rotated_binary_search(rotated_dataset, target_incident)
lin_idx, lin_comp = linear_search(rotated_dataset, target_incident)

# 3. Assessment Output
print("--- Exercise 1 Output ---")
print("Original Dataset:       ", original_dataset)
print("Rotated Dataset:        ", rotated_dataset)
print("Target Incident:        ", target_incident)
print(f"Rotated Search Result:   Index {bin_idx}, Comparisons = {bin_comp}")
print(f"Linear Search Result:    Index {lin_idx}, Comparisons = {lin_comp}")
print("Time Complexity:         O(log n) for Rotated Binary, O(n) for Linear")
print("Space Complexity:        O(1)")
```

```python
# Exercise 2: University Enrollment Records

def merge_enrollments(dept_a, dept_b):
    i, j = 0, 0
    consolidated = []
    comparisons = 0
    
    while i < len(dept_a) and j < len(dept_b):
        comparisons += 1
        if dept_a[i] <= dept_b[j]:
            consolidated.append(dept_a[i])
            i += 1
        else:
            consolidated.append(dept_b[j])
            j += 1
            
    while i < len(dept_a):
        consolidated.append(dept_a[i])
        i += 1
    while j < len(dept_b):
        consolidated.append(dept_b[j])
        j += 1
        
    return consolidated, comparisons

# 1. Dataset Generation
dept_a = sorted([(1000 + SEED_DIGIT * 10 + i * ID_SUM) for i in range(8)])
dept_b = sorted([(1000 + ARTIST_LENGTH * 5 + i * NAME_LENGTH) for i in range(7)])

# 2. Consolidation Execution
consolidated_list, merge_comparisons = merge_enrollments(dept_a, dept_b)
is_sorted = consolidated_list == sorted(consolidated_list)

# 3. Assessment Output
print("\n--- Exercise 2 Output ---")
print("Department A Dataset:   ", dept_a)
print("Department B Dataset:   ", dept_b)
print("Consolidated Dataset:   ", consolidated_list)
print(f"Records Before/After:   {len(dept_a) + len(dept_b)} -> {len(consolidated_list)}")
print("Comparison Count:       ", merge_comparisons)
print("Ordering Verification:  ", "Successfully Sorted" if is_sorted else "Failed")
print("Time Complexity:         O(n + m)")
print("Space Complexity:        O(n + m)")
```

```python
# Exercise 3: E-Commerce Order Processing

import random
import time

def bubble_sort(data):
    arr = data.copy()
    n = len(arr)
    comp, swaps = 0, 0
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            comp += 1
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return arr, comp, swaps

def selection_sort(data):
    arr = data.copy()
    n = len(arr)
    comp, swaps = 0, 0
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comp += 1
            if arr[j] < arr[min_idx]:
                min_idx = j
        if min_idx != i:
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
            swaps += 1
    return arr, comp, swaps

def insertion_sort(data):
    arr = data.copy()
    comp, shifts = 0, 0
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0:
            comp += 1
            if arr[j] > key:
                arr[j + 1] = arr[j]
                shifts += 1
                j -= 1
            else:
                break
        arr[j + 1] = key
    return arr, comp, shifts

def measure_sort(sort_fn, data):
    start = time.perf_counter()
    res, comp, moves = sort_fn(data)
    elapsed = time.perf_counter() - start
    return res, comp, moves, elapsed

# 1. Dataset Generation
base_orders = [(ID_SUM * 2 + i * 5) for i in range(20)]
sorted_orders = sorted(base_orders)
reverse_orders = list(reversed(sorted_orders))
random_orders = base_orders.copy()
random.seed(SEED_DIGIT)
random.shuffle(random_orders)

conditions = {
    "Sorted": sorted_orders,
    "Reverse": reverse_orders,
    "Random": random_orders
}

algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

# 2. Performance Analysis Execution
print("\n--- Exercise 3 Output ---")
print(f"{'Condition':<10} | {'Algorithm':<15} | {'Comparisons':<12} | {'Movements':<10} | {'Time (s)':<12}")
print("-" * 70)

for cond_name, dataset in conditions.items():
    for alg_name, alg_func in algorithms.items():
        res, comp, moves, elapsed = measure_sort(alg_func, dataset)
        assert res == sorted(dataset), f"{alg_name} failed sorting!"
        print(f"{cond_name:<10} | {alg_name:<15} | {comp:<12} | {moves:<10} | {elapsed:<12.8f}")
```
